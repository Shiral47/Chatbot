# Creative Uses for Everyday Objects

This notebook asks for an everyday object and uses OpenAI to suggest one unusual, creative, and practical way to use it.

You will need an OpenAI API key at runtime. The key is entered privately and is not saved in this notebook.

## 1. Install the OpenAI SDK if needed

In a Colab code cell, the install command is `!pip -q install openai`. The next cell checks for the SDK first and installs it only when it is missing.

```python
import importlib.util
import subprocess
import sys

if importlib.util.find_spec("openai") is None:
    # Install the SDK into this Colab runtime if it is not already installed.
    subprocess.check_call([sys.executable, "-m", "pip", "-q", "install", "openai"])

from openai import OpenAI
from getpass import getpass
```

## 2. Enter your API key securely

Get an API key from your OpenAI account. The prompt hides what you type. Do not paste your key into notebook code or save it in a notebook output.

```python
# getpass keeps the key hidden while you type it.
api_key = getpass("Enter your OpenAI API key: ")

# Create a client for this runtime only; the key is not written to a file.
client = OpenAI(api_key=api_key)
```

## 3. Choose an everyday object

Type a common object, such as a paper clip, mug, or shoelace.

```python
object_name = input("Enter an everyday object: ").strip()

if not object_name:
    raise ValueError("Please enter an object before continuing.")
```

## 4. Generate one creative idea

The Responses API will return one unusual, creative, and practical use for your object.

```python
response = client.responses.create(
    model="gpt-6-luna",
    input=f"Suggest exactly ONE unusual, creative, and practical use for this everyday object: {object_name}. Keep it concise and explain how to use it.",
)

print(response.output_text)
```
